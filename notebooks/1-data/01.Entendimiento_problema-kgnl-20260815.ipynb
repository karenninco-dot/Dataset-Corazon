## Preguntas- Respuestas

- ¿Cual es el objetivo del problema?
    
    El objetivo general es desarrollar un sistema basado en aprendizaje automático que permita determinar automáticamente si un paciente tiene problemas del corazón a partir de los datos obtenidos de sus exámenes y características clínicas.

- ¿Cómo se usará su solución?
    
    La solución se utilizaría como un sistema de apoyo para determinar automáticamente si un paciente presenta o no problemas del corazón, utilizando como entrada los datos disponibles de sus exámenes.

    La salida esperada sería:

    1 → el paciente tiene enfermedad/problema cardíaco.
    0 → el paciente no tiene enfermedad/problema cardíaco.

- ¿Cuáles son las soluciones actuales (si las hay)?
    
    No se documenta una solución actual específica; el proyecto busca desarrollar una solución automatizada mediante Machine Learning.

- ¿Cómo se debe enmarcar este problema (supervisado / no supervisado, en línea / fuera de línea, etc.)
    
    Es principalmente un problema de Aprendizaje supervisado → clasificación binaria.

    Tenemos:
    Variables de entrada: edad, sexo, presión arterial, colesterol, etc.
    Una variable objetivo conocida: disease.
    La variable objetivo tiene dos resultados principales:
    0: no enfermedad.
    1: enfermedad.

- ¿Cómo se debe medir el desempeño o el rendimiento de la solución, una primera intuicion?
    
    Las métricas que inicialmente consideraría son:
    Recall: qué proporción de los pacientes que realmente tienen enfermedad cardíaca son detectados.
    Precision: de los pacientes que el modelo identifica como enfermos, cuántos realmente lo están.
    F1-score: equilibrio entre precision y recall.
    Matriz de confusión: para analizar falsos positivos y falsos negativos.
    Accuracy: como métrica complementaria.

    La métrica que merece especial atención es el Recall, porque un falso negativo —decir que un paciente no tiene enfermedad cuando sí la tiene— podría ser especialmente problemático.

- ¿La medida de desempeño está alineada con el objetivo del problema?
    
    Sí, si se prioriza Recall/Sensibilidad.
    El objetivo es detectar pacientes que puedan presentar problemas cardíacos. Por lo tanto, tiene sentido minimizar los casos en los que:
    El paciente realmente tiene la enfermedad → pero el sistema predice que no.
    Es decir, se debe prestar especial atención a los falsos negativos.
    Sin embargo, no sería correcto decir que solamente importa Recall. También debemos controlar los falsos positivos y, por tanto, observar Precision y la matriz de confusión.

- ¿Cuál sería el desempeño o rendimiento mínimo necesario para alcanzar el objetivo del problema?
    
    En esta etapa podríamos establecer como supuesto inicial que: El modelo debe alcanzar un desempeño suficientemente alto para que la detección de pacientes con posible enfermedad cardíaca sea confiable, dando especial importancia a minimizar los falsos negativos.
